Question 1 : What is Dimensionality Reduction? Why is it important in machine learning?

- Dimensionality Reduction is the process of reducing the number of input variables/features while preserving as much important information as possible.

It is important because:

  - It removes noise and redundant features, improving model performance.

  - It reduces computational cost and storage requirements.

- It helps visualize high-dimensional data in 2D or 3D.

- It reduces the risk of overfitting.

Question 2: Name and briefly describe three common dimensionality reduction techniques.

Answer:

- PCA (Principal Component Analysis):
Converts original features into a smaller set of uncorrelated components that capture maximum variance.

 - t-SNE (t-Distributed Stochastic Neighbor Embedding):
A nonlinear technique mainly used for visualization by preserving local structure of high-dimensional data.

- LDA (Linear Discriminant Analysis):
A supervised technique that reduces dimensions while maximizing class separability.

Question 4: Explain the concept of anomaly detection and its significance.

Answer:
- Anomaly Detection is the process of identifying data points that significantly differ from the normal pattern or behavior.
Significance:

- Helps detect fraud, network intrusions, machine failures, and health abnormalities.

- Improves system reliability and security by alerting unusual activities early.

Question 5: List and briefly describe three types of anomaly detection techniques.

Answer:

- Statistical Methods:
Identify anomalies using probability distributions, mean, variance, Z-scores, etc.

- Machine Learning–Based Methods:
Use models like Isolation Forest, One-Class SVM, or Autoencoders to learn normal patterns and detect deviations.

- Distance- or Density-Based Methods:
Detect outliers based on distance from neighbors (e.g., k-NN) or density (e.g., LOF – Local Outlier Factor).

Question 6: What is time series analysis? Mention two key components of time series data.

Answer:
- Time Series Analysis is the study of data points collected or recorded in time order to understand trends, patterns, and future predictions.
Key components:

- Trend: Long-term upward or downward movement in the data.

- Seasonality: Repeating patterns or cycles that occur at regular intervals (daily, monthly, yearly).

Question 7: Describe the difference between seasonality and cyclic behavior in time series.

Answer:

- Seasonality:

   - Refers to patterns that repeat at fixed, regular intervals (daily, monthly, yearly).

   - Caused by predictable factors like weather, holidays, or routines.

  - Example: Ice-cream sales increase every summer.

- Cyclic Behavior:

  - Refers to patterns that occur at irregular, non-fixed intervals.

  - The duration of a cycle is not consistent, influenced by economic or natural factors.

  - Example: Business cycles (expansion and recession).

  Question 8: Python code to perform K-means clustering on a sample dataset

```python
import numpy as np
from sklearn.cluster import KMeans

X = np.array([[1, 2],
              [1, 4],
              [1, 0],
              [10, 2],
              [10, 4],
              [10, 0]])

kmeans = KMeans(n_clusters=2, random_state=0, n_init=10).fit(X)
labels = kmeans.labels_
centroids = kmeans.cluster_centers_
labels, centroids
```

Question 9: What is inheritance in OOP? Provide a simple example in Python.

Answer:
- Inheritance is an OOP concept where a new class (child/subclass) acquires the properties and methods of an existing class (parent/superclass).
It promotes code reuse and modularity.

```python
class Animal:
    def speak(self):
        return "Animal makes a sound"

class Dog(Animal):   # Dog inherits from Animal
    def speak(self):
        return "Dog barks"

d = Dog()
print(d.speak())
```

Question 10: How can time series analysis be used for anomaly detection?

Answer:
- Time series analysis helps detect anomalies by identifying data points that deviate from expected temporal patterns. Methods include:

   - Forecasting models (ARIMA, LSTM): Compare predicted values with actual values; large errors indicate anomalies.

  - Seasonality and trend decomposition: Points that break normal seasonal or trend patterns are flagged.

   - Statistical thresholds: Using moving averages, Z-scores, or confidence intervals to detect outliers.

It is widely used in fraud detection, network monitoring, sensor failure detection, and financial anomaly detection.
